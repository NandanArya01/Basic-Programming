##Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
number = int(input("Enter a number: "))
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
if is_prime(number):
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

#Product of Random Numbers

```python
import random

def generate_numbers():
    num1 = random.randint(1, 10)
    num2 = random.randint(1, 10)
    return num1, num2

def main():
    num1, num2 = generate_numbers()
    correct_sum = num1 + num2

    user_answer = int(input(f"What is the sum of,{num1} and {num2}? "))


    if user_answer == correct_sum:
        print("Your answer is Correct.")
    else:
        print(f"Incorrect. The correct answer is {correct_sum}.")

if __name__ == "__main__":
    main()
```

#Squares of all even numbers within the range of 100 to 200


```python

def print_squares_of_even_numbers():
    for num in range(100, 201):
        if num % 2 == 0:
            print(f"The square of {num} is {num ** 2}")
print_squares_of_even_numbers()
```

```python
def print_cube_of_even_numbers():
    for num in range(100, 201):
        if num % 2 == 0:
            print(f"The cube of {num} is {num ** 3}")
print_cube_of_even_numbers()
```

#Word counter

```python
def count_words(input_text):
    input_text = input_text.lower()
    words = input_text.split()
    word_counts = {}
    for word in words:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1
    return word_counts
input_text = "This is a sample text. This text will be used to demonstrate the word counter."

word_counts = count_words(input_text)
for word, count in word_counts.items():
    print(f"'{word}': {count}")
```

#Check for Palindrome

```python
import re

def is_palindrome(s):
    s = re.sub(r'[^A-Za-z0-9]', '', s).lower()
    return s == s[::-1]
input_string = "A man, a plan, a canal, Panama"
print(f"Input: \"{input_string}\"")
print(f"Expected Output: {is_palindrome(input_string)}")
```
